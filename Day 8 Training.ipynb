```python
useraccount = {'username' : 'codetraining', 'firstname' : 'python', 'lastname' : 'training','dob' : 'sept 30', 'pwd' : '1234'}

for k,v in useraccount.items():
  print(k)
  print(v)
```

```python
useraccount = {'username' : 'codetraining', 'firstname' : 'python', 'lastname' : 'training','dob' : 'sept 30', 'pwd' : '1234'}

for k,v in useraccount.items():
  print(f"key: {k}")
  print(f"key: {v}")
```

```python
useraccount = {'username' : 'codetraining', 'firstname' : 'python', 'lastname' : 'training','dob' : 'sept 30', 'pwd' : '1234'}

for k,v in useraccount.items():
  print(f"key: {k}")
  print(f"key: {v}\n")
```

```python
useraccount = {'username' : 'codetraining', 'firstname' : 'python', 'lastname' : 'training','dob' : 'sept 30', 'pwd' : '1234'}

for k in useraccount.keys():
  print(f"key: {k}")
```

```python
useraccount = {'username' : 'codetraining', 'firstname' : 'python', 'lastname' : 'training','dob' : 'sept 30', 'pwd' : '1234'}

for v in useraccount.values():
  print(f"value: {v}")
```
